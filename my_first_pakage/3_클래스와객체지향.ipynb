```python
# 사용자 정의 예외 클래스 만들기
class AgeException(Exception):
    pass


# 나이를 입력받는 함수
def input_age():

    # 나이 입력
    age = int(input("나이를 입력하세요 : "))

    # 입력값 체크

    # 1. 나이가 0보다 작으면 예외 발생
    if age < 0:
        raise AgeException("나이는 0보다 작을 수 없습니다.")

    # 2. 나이가 101보다 크면 예외 발생
    elif age > 101:
        raise AgeException("나이가 너무 큽니다.")

    # 3. 정상적인 나이라면 반환
    else:
        return age


# 예외처리를 하는 나이 입력 구현
try :
    age = input_age()
except AgeException as e:
    print(e.args[0])
else :
    print(f'입력하신 나이는 {age}입니다.')
```
